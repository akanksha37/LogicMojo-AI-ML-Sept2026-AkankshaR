```python
# Q29. Two models use the following input features:
#
# model_a_features = {"age", "salary", "tenure",
#                     "credit_score", "balance"}
#
# model_b_features = {"age", "salary", "tenure",
#                     "department", "region"}
#
# Find the common features and the features unique to each
# model.
# Calculate the Jaccard similarity between the feature sets.
# Formula:
# Jaccard Similarity = Size of Intersection / Size of Union

model_a_features = {"age", "salary", "tenure", "credit_score", "balance"}
model_b_features = {"age", "salary", "tenure","department", "region"}
print(model_a_features)
print(model_b_features)
common = model_a_features.intersection(model_b_features)
print(common)
unique_to_a = model_a_features - model_b_features
print(unique_to_a)
unique_to_b = model_b_features - model_a_features
print(unique_to_b)
intersec_size = len(common)
print(intersec_size)
union_size = len(model_a_features.union(model_b_features))
print(union_size)
jaccard_similarity = intersec_size/union_size
print(f"Jaccard Similarity: {jaccard_similarity:.2f}")
```

```python
# Q37. A data scientist stores the following model results:
#
# model_results = {
#     "Linear Regression": {"accuracy": 0.82, "f1_score": 0.80},
#     "Decision Tree": {"accuracy": 0.88, "f1_score": 0.86},
#     "Random Forest": {"accuracy": 0.94, "f1_score": 0.92}
# }
#
# Print the accuracy and F1-score of each model.
# Identify the model with the highest accuracy.
# Print its name and accuracy.

model_results = {
    "Linear Regression": {"accuracy": 0.82, "f1_score": 0.80},
    "Decision Tree": {"accuracy": 0.88, "f1_score": 0.86},
    "Random Forest": {"accuracy": 0.94, "f1_score": 0.92}
}
highest_accuracy = -1.0
best_model_name = ""
for model, metrics in model_results.items():
    # Access inner dictionary values using two keys: metrics["accuracy"] and metrics["f1_score"]
    acc = metrics["accuracy"]
    f1 = metrics["f1_score"]
    print(f"{model:<17} | Accuracy: {acc} | F1-Score: {f1}")
    if acc > highest_accuracy:
        highest_accuracy = acc
        best_model_name = model
print(f"The model with the highest accuracy is '{best_model_name}' with a score of {highest_accuracy}.")
```

```python
# Q39. A dataset contains the following actual customer
# categories:
#
# actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]
#
# Create a dictionary that counts the frequency of each
# category.
#
# Then calculate the percentage of records belonging to
# each category.
#
# Example:
# Category A: 37.5%
#
# Print both the counts and percentages.

actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]
category_counts = {}
for category in actual_categories:
    category_counts[category] = category_counts.get(category, 0) + 1
total_records = len(actual_categories)
for cat, count in category_counts.items():
    print(f"Category {cat}: Count = {count}")
for cat, count in category_counts.items():
    # Calculate percentage
    percentage = (count / total_records) * 100
    print(f"Category {cat}: {percentage:.1f}%")
```

```python
# Q40. A data science team maintains two dictionaries
# containing model performance results:
#
# experiment_1 = {
#     "Linear Regression": 0.82,
#     "Decision Tree": 0.88,
#     "Random Forest": 0.94
# }
#
# experiment_2 = {
#     "Decision Tree": 0.90,
#     "Random Forest": 0.96,
#     "KNN": 0.85
# }
#
# Perform the following tasks:
#
# 1. Merge both dictionaries into a new dictionary.
# 2. If a model appears in both dictionaries, retain the
#    higher accuracy score.
# 3. Print the final dictionary.
# 4. Identify the model with the highest accuracy.
# 5. Print all models whose final accuracy is greater than 0.90.

experiment_1 = {
    "Linear Regression": 0.82,
    "Decision Tree": 0.88,
    "Random Forest": 0.94
}

experiment_2 = {
    "Decision Tree": 0.90,
    "Random Forest": 0.96,
    "KNN": 0.85
}
final_results = experiment_1.copy()
for model, score in experiment_2.items():
    if model in final_results:
        final_results[model] = max(final_results[model], score)
    else:
        final_results[model] = score
print(final_results)
best_model = max(final_results, key = final_results.get)
print(f"Model with the Highest Accuracy: '{best_model}' ({final_results[best_model]})")
for model, score in final_results.items():
    if score > 0.90:
        print(f"- {model}: {score}")
```
