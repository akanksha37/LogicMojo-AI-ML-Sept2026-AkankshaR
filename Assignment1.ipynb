```python
#42. Take 10 prediction errors as input using a loop. Calculate and print their mean absolute error (MAE) without using lists or libraries.

total_absolute_error = 0.0
total_predictions = 10
for i in range(1, total_predictions + 1):
    error = float(input(f"Enter prediction error {i}: "))
    if error < 0:
        absolute_error = -error
    else:
        absolute_error = error
    total_absolute_error += absolute_error
mae = total_absolute_error / total_predictions
print(f"\nThe Mean Absolute Error (MAE) is: {mae}")
```

```python
#43. Take 10 actual values and 10 predicted values, one pair at a time. Calculate the mean squared error (MSE) using a loop.

total_squared_error = 0.0
total_pairs = 10
for i in range(1, total_pairs + 1):
    actual = float(input(f"Enter actual value {i}: "))
    predicted = float(input(f"Enter predicted value {i}: "))
    error = actual - predicted
    squared_error = error * error
    total_squared_error += squared_error
mse = total_squared_error / total_pairs
print(f"\nThe Mean Squared Error (MSE) is: {mse}")
```

```python
#44. Write a function calculate_mae(error1, error2, error3) that returns the average absolute error for three predictions.
def calculate_mae(error1, error2, error3):
    abs1 = -error1 if error1 < 0 else error1
    abs2 = -error2 if error2 < 0 else error2
    abs3 = -error3 if error3 < 0 else error3
    mae = (abs1 + abs2 + abs3) / 3
    return mae
result = calculate_mae(-2.5, 4.0, -1.5)
print(result)
```

```python
#45. Write a function that accepts correct predictions and total predictions, calculates accuracy, and classifies the model as "Excellent" (at least 95%), "Good" (at least 85%), or "Needs Improvement".

def evaluate_model(correct, total):
    if total == 0:
        return 0.0, "Needs Improvement"
    accuracy = correct / total
    if accuracy >= 0.95:
        classification = "Excellent"
    elif accuracy >= 0.85:
        classification = "Good"
    else:
        classification = "Needs Improvement"
    return accuracy, classification
acc, rating = evaluate_model(96, 100)
print(f"Accuracy: {acc * 100}%, Classification: {rating}")
```

```python
#51. Write a function that accepts any number of validation scores using *args and returns their average. Then use it to compare two model averages with a conditional statement.

def calculate_average(*args):
    if not args:
        return 0.0
    total_sum = 0.0
    count = 0
    for score in args:
        total_sum += score
        count += 1
    return total_sum / count
model_a_avg = calculate_average(0.85, 0.88, 0.90, 0.87)
model_b_avg = calculate_average(0.82, 0.91, 0.89, 0.86)
print(f"Model A Average: {model_a_avg:.4f}")
print(f"Model B Average: {model_b_avg:.4f}")
if model_a_avg > model_b_avg:
    print("Model A has a higher average validation score.")
elif model_b_avg > model_a_avg:
    print("Model B has a higher average validation score.")
else:
    print("Both models have the exact same average validation score.")
```

```python
#55. Write a decorator that prints "Training started" before a function executes and "Training finished" after it executes. Apply it to a function that prints the current epoch.

def training_decorator(func):
    def wrapper(epoch):
        print("Training started")
        func(epoch)
        print("Training finished")
    return wrapper
@training_decorator
def show_epoch(epoch):
    print("Current epoch:", epoch)
show_epoch(5)
```

```python
#56. Write a decorator that accepts a function, executes it, and returns its result. Demonstrate it using a function that calculates model accuracy.

def result_decorator(func):
    def wrapper(correct, total):
        result = func(correct, total)
        return result
    return wrapper
@result_decorator
def calculate_accuracy(correct, total):
    return correct / total
accuracy = calculate_accuracy(90, 100)
print("Accuracy:", accuracy)
```

```python
#57. Create a lambda function that calculates squared error from actual and predicted values. Use it inside a regular function to calculate the total squared error for three predictions.

squared_error = lambda actual, predicted: (actual - predicted) ** 2
def total_squared_error(a1, p1, a2, p2, a3, p3):
    error1 = squared_error(a1, p1)
    error2 = squared_error(a2, p2)
    error3 = squared_error(a3, p3)
    return error1 + error2 + error3
print(total_squared_error(10, 8, 20, 18, 30, 27))
```

```python
#60. Build a mini training simulation using functions, a decorator, a loop, conditions, break, and a lambda function. Print the loss for each epoch, stop when loss reaches 0.10 or lower, and return the final loss. Use only the Python concepts covered in class.

reduce_loss = lambda loss: loss - 0.1
def training_decorator(func):
    def wrapper():
        print("Training started")
        result = func()
        print("Training finished")
        return result
    return wrapper
@training_decorator
def train_model():
    loss = 1.0
    for epoch in range(1, 20):
        loss = reduce_loss(loss)
        print("Epoch:", epoch, "Loss:", round(loss, 2))
        if loss <= 0.10:
            break
    return loss
final_loss = train_model()
print("Final Loss:", round(final_loss, 2))
```
